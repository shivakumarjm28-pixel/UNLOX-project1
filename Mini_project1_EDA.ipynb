```python
# visulization
numarical_column=df.select_dtypes(include=['int64','float64']).columns
object_column=df.select_dtypes(include=['object']).columns
```

```python
for i in numarical_column:
  plt.figure(figsize=(6,4))
  sns.distplot(x=df[i])
  plt.title(f"boxplot of {i}")
  plt.show()
```

```python
for col in object_column:
  plt.figure(figsize=(6,4))
  sns.countplot(x = df[col])
  plt.title(f"boxplot of {col}")
  plt.show()
```

```python
#univariate analysis
plt.figure(figsize=(8,5))
sns.histplot(data = df,y=  "Pclass")
plt.title("price distribution")
plt.show()
```

```python
#Bi variate
plt.figure(figsize=(8,5))
sns.scatterplot(x= df["Pclass"],y=df["SibSp"])
plt.title("pclass vs SibSp")
plt.show()
```

```python
#multi
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True), annot=True,cmap="Reds")

plt.show()
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Fare']=df['Fare'].fillna(df['Fare'].median())
```

```python
df['Ticket']=df['Ticket'].fillna(df['Ticket'].mode()[0])
```

```python
#outlier detect
plt.figure(figsize=(8,5))
sns.boxplot(x=  df["Pclass"])
plt.title("pclass distribution")
plt.show()
```

```python
for col in numarical_column:
  plt.figure(figsize=(8,5))
  sns.boxplot(x=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
#outlier treatement
for col in numarical_column:
  q1=df[col].quantile(0.25)
  q3=df[col].quantile(0.75)
  iqr=q3-q1
  lowr_limit=q1-1.5*iqr
  upper_limit=q3+1.5*iqr
  df[col]=np.where(df[col]>upper_limit,upper_limit,df[col])
```

```python
numarical_column=df.select_dtypes(include=['int64','float64']).columns
for col in numarical_column:
  plt.figure(figsize=(8,5))
  sns.boxplot(x=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
#check unque
for col in df.select_dtypes(include=['object']).columns:
  print(col)
  print(df[col].unique())
```

```python
for col in df.select_dtypes(include=['object']).columns:
    df[col]=df[col].str.strip().str.title()
```

```python
for col in df.select_dtypes(include=['object']).columns:
  print(col)
  print(df[col].unique())
```

```python
#lable encoding
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for col in df.select_dtypes(include=['object']).columns:
  df[col]=le.fit_transform(df[col])
df.head()
```
